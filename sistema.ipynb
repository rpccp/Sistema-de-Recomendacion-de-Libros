# Sistema de Recomendacion de Libros

Demo del recomendador basado en embeddings, Ollama y FAISS. La logica principal vive en `src/recommender.py` para que el proyecto pueda ejecutarse desde notebook o desde la CLI (`app.py`).

## 1. Cargar datos e indice

Antes de ejecutar el notebook, asegurate de tener Ollama corriendo y el modelo instalado:

```bash
ollama serve
ollama pull deepseek-r1:7b
```

````python
from pathlib import Path

from src.recommender import BookRecommender, load_books

DATA_PATH = Path("books.csv")
INDEX_PATH = Path("indice")
MODEL = "deepseek-r1:7b"
````

````python
books = load_books(DATA_PATH)
books[["title", "authors", "categories", "average_rating"]].head()
````

## 2. Buscar libros por titulo

````python
books[books["title"].str.contains("Rich", case=False, na=False)][[
    "title", "authors", "categories", "average_rating"
]].head(10)
````

## 3. Recomendar libros similares

````python
recommender = BookRecommender(
    data_path=DATA_PATH,
    index_path=INDEX_PATH,
    model=MODEL,
)
````

````python
recommendations = recommender.recommend("Rich", k=5)

for book in recommendations:
    print(book.title)
    print(f"Autor/es: {book.authors}")
    print(f"Categoria: {book.categories}")
    print(f"Rating: {book.average_rating}")
    print(f"Distancia FAISS: {book.distance:.4f}")
    print()
````
